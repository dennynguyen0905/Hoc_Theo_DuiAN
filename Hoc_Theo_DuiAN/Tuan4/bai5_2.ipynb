# Bài 2: Dùng next()
Sử dụng next() để lấy 2 giá trị đầu tiên từ generator ở 
Bài 1

```python
gen = (x for x in range(16) if x % 3 == 0)

first = next(gen)
second = next(gen)
print("Số thứ 1: ", first)
print("Số thứ 2: ", second)
```

## Bai 3 generator  next()

```python
def is_prime(x):
    if x < 2:
        return 0
    for i in range(2, int(x**0.5) + 1):
        if x % i == 0:
            return 0
    return 1 

prime = (x for x in range(1, 21) if is_prime(x))

try:
    while 1:
        print(next(prime))
except StopIteration:
    pass
```
