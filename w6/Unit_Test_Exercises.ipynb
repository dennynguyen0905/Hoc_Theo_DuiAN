### **Bài 1. Viết test cho hàm tính thuế:**

```python
def calculate_tax(income):
    if income < 5000:
        return 0
    elif income < 10000:
        return income * 0.1
    else:
        return income * 0.2

# TODO: Viết test cho các trường hợp: income < 5000, income = 7000, income = 12000

# Kiểm tra các trường hợp
assert calculate_tax(3000) == 0, "Lỗi ở mức thu nhập thấp"
assert calculate_tax(7000) == 700.0, "Lỗi ở mức thu nhập trung bình"
assert calculate_tax(12000) == 2400.0, "Lỗi ở mức thu nhập cao"

print("Tất cả các test case đều vượt qua!")
```

### **Bài 2. Viết test cho hàm kiểm tra số nguyên tố:**

```python
def is_prime(n):
    if n <= 1: return False
    for i in range(2, int(n**0.5)+1):
        if n % i == 0:
            return False
    return True

# TODO: Viết test cho các số: 1, 2, 3, 4, 17, 18, 19
assert is_prime(1) == False, "Lỗi ở số 1"
assert is_prime(2) == True, "Lỗi ở số 2"
assert is_prime(3) == True, "Lỗi ở số 3"        
assert is_prime(4) == False, "Lỗi ở số 4"
assert is_prime(17) == True, "Lỗi ở số 17"
assert is_prime(18) == False, "Lỗi ở số 18"
assert is_prime(19) == True, "Lỗi ở số 19"
```

### **Bài 4. Mock hàm gửi email:**

```python
import unittest
from unittest.mock import patch

def send_welcome_email(email):
    print(f"Sending email to {email}")

class TestEmailService(unittest.TestCase):
    
    @patch('builtins.print')  # Thay thế print bằng một mock object
    def test_send_welcome_email(self, mock_print):
        email_input = "linhhai@gmail.com"
        send_welcome_email(email_input)
        
        mock_print.assert_called()
        
        self.assertEqual(mock_print.call_count, 1)
        
        mock_print.assert_called_once_with(f"Sending email to {email_input}")

unittest.main(argv=['first-arg-is-ignored'], exit=False)
```

### **Bài 5. Mock gọi API với requests.get:**

```python
import requests

def fetch_user():
    response = requests.get("https://api.example.com/user")
    return response.json()

# TODO: Dùng unittest.mock để giả lập kết quả requests.get()

import unittest
from unittest.mock import patch, Mock
class TestFetchUser(unittest.TestCase):
    
    @patch('requests.get')
    def test_fetch_user(self, mock_get):
        mock_response = Mock()
        expected_data = {"name": "John Doe", "age": 30}
        mock_response.json.return_value = expected_data
        mock_get.return_value = mock_response
        
        result = fetch_user()
        
        self.assertEqual(result, expected_data)
        mock_get.assert_called_once_with("https://api.example.com/user")
```

### **Bài 6. Mock ngày hiện tại để test cuối tuần:**

```python
from datetime import datetime

def is_weekend():
    today = datetime.now().weekday()
    return today >= 5

# TODO: Mock datetime.now() để test với thứ 6 (4) và thứ 7 (5)
import unittest
from unittest.mock import patch
class TestIsWeekend(unittest.TestCase):
    
    @patch('datetime.datetime')
    def test_is_weekend(self, mock_datetime):
        mock_datetime.now.return_value.weekday.return_value = 4
        self.assertFalse(is_weekend())
        
        mock_datetime.now.return_value.weekday.return_value = 5
        self.assertTrue(is_weekend())

```

### **Bài 7. Viết test cho hàm kiểm tra mật khẩu:**

```python
def is_strong(password):
    return len(password) >= 8 and any(c.isdigit() for c in password)

# TODO: Test với các password yếu và mạnh, kiểm tra đúng/false
assert is_strong("112") == False, "Lỗi ở password yếu"
assert is_strong("pass1234") == True, "Lỗi ở password mạnh"
```

### **Bài 9. Viết test cho hàm chia số:**

```python
def safe_divide(a, b):
    if b == 0:
        return None
    return a / b

# TODO: Test các trường hợp: chia 10/2, 5/0, -4/2
assert safe_divide(10, 2) == 5.0, "Lỗi ở phép chia bình thường"
assert safe_divide(5, 0) == None, "Lỗi ở phép chia cho 0"
assert safe_divide(-4, 2) == -2.0, "Lỗi ở phép chia với số âm"
```
