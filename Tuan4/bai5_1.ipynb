```python
class Book:
    def __init__(self, title, author, year_publish):
        self.title = title
        self.author = author
        self.year_publish = year_publish
        
    def  display_info (self):
        print(f"Tên sách: {self.title}.")
        print(f"Tác giả: {self.author}.")
        print(f"Năm xuất bản: {self.year_publish}")
        print("-" * 10)
        
book1 = Book("Deep work", "Cal Newport", 2012)       
book2 = Book("Lão Hạc", "Nam Cao", 1943)    

book1.display_info()
book2.display_info()
```

## Bài 2

```python
class BankAccount():
    def __init__(self, account_Number, balance=0):
        self._account_Number = account_Number
        self._balance = balance
    
    @property
    def account_Number(self):
        return self._account_Number
        
    def deposit(self, amount):
        if amount > 0:
            self._balance += amount
            print(f'Nạp thành công {amount}, số dư của bạn là  {self._balance}')
        else:
            print("Số tiền nạp vào phải lớn hơn 0")
            
    def withdraw(self, amount):
        if amount <= 0:
            print("Số tiền rút phải lớn hơn 0")
        elif amount > self._balance:
            print("Số dư tài khoản không đủ!")
        else:
            self._balance -= amount
            print(f"Rút thành công {amount}, số dư tài khoản  {self._balance}")
            
    def get_balance(self):
        return self._balance
    
account1 = BankAccount("9090905210906")

account1.withdraw(90000)
account1.deposit(1000000)
account1.withdraw(430000)

print(f"Tai khoan {account1._account_Number} co so du la  ", account1.get_balance())
print("Số tài khoản  ", account1.account_Number)
```

## Bai 3

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
        
    def display_info(self):
        print(f"Sinh viên {self.name},")
        print(f"{self.age} tuổi.")
        print("--" * 10)
        
class student(Person):
    def __init__(self, name, age, studentID):
        super().__init__(name, age)
        self.studentID = studentID
        
    def study(self):
        print(f"Sinh viên {self.name} đang học bài")
        
student1 = student("Nguyễn Quốc Khánh", 21, "BCSE2036")

student1.display_info()
student1.study()

print("Mã số sinh viên ", student1.studentID)
print("Tên", student1.name)
```

## Bai 4

```python
class Vehicle:
    def __init__(self, brand, model, year):
        self.brand = brand
        self.model = model
        self.year = year
        
    def display_info(self):
        print(f"Xe hãng {self.brand}")
        print(f"Model {self.model}")
        print(f"Sản xuất {self.year}")
        
class Car(Vehicle):
    def __init__(self, brand, model, year, num_doors):
        super().__init__(brand, model, year)
        self.num_doors = num_doors
        
    def display_info(self):
        super().display_info()
        print(f"Số cửa {self.num_doors}")
        print("--" * 10)
        
class Bike(Vehicle):
    def __init__(self, brand, model, year, type):
        super().__init__(brand, model, year)
        self.type = type

    def display_info(self):
        super().display_info()    
        print(f"Loại xe {self.type}")
        
car1 = Car("Vinfast", "LimoGreen", 2025, 4)
bike1 = Bike("Honda", "Air Blade 160", 2026, "Sport")

print("thông tin ô tô: ")
car1.display_info()

print("\nThông tin xe máy: ")
bike1.display_info()
```
